## Reverse: Stack

1. Create a function that reverses individual characters in a stack. Note that the original stack should be reversed.

2. Add a `reverse()` method to the `ListStack` class.

```python
# A function that reverses individual characters in a stack. Note that the original stack should be reversed.
def reverse(stack):
   # FILL OUT HERE
    reverse_stack = ListStack()
    stack2 = ListStack()

    while not stack.isEmpty():
      reverse_stack.push(stack.pop())

    while not reverse_stack.isEmpty():
      stack2.push(reverse_stack.pop())

    while not stack2.isEmpty():
        stack.push(stack2.pop())

    return   stack.printStack()


stack = ListStack()
stack.push(100); stack.push(200); stack.push(300); stack.push(400)
print("Original Stack: Before Reversing")
stack.printStack()

print("Reversed Stack: ")
reverse(stack)

print("Original Stack: After Reversing")
stack.printStack()
```

## Reverse: Queue

1. Create a function that reverses individual characters in a queue using recursive approach. Note that the original queue should be reversed.
2. Try to add `reverse()` method to the ListQueue class.

```python
from DataStructureCode.LinkedQueue import *
queue = ListQueue()
queue.enqueue("Mon"); queue.enqueue("Tue"); queue.enqueue("Wed"); queue.enqueue("Thurs"); queue.enqueue("Fri")
queue.printQueue()

# A function that reverses individual characters in a queue using recursive approach. Note that the original queue should be reversed.
def reverseQueue(queue):
    # FILL OUT HERE
    if queue.isEmpty():
        return

    item = queue.dequeue()
    reverseQueue(queue)
    queue.enqueue(item)

reverseQueue(queue)
print("After Reversing the queue:")
queue.printQueue()
```

```python
# add the below method to the ListQueue class.
def reverse(self):
		# FILL OUT HERE

		if self.isEmpty():
			return

		item = self.dequeue()
		self.reverse()
		self.enqueue(item)
```

## Palindrome

*   Palindrome is word, phrase, or sequence that reads the same backward as forward (e.g. madam, mom, kayak). Write a function that checks if a string is palindrome using stack and queue data structure.



```python
from DataStructureCode.ListStack import *
from DataStructureCode.ListQueue import *

def isPalindrome(A):
    # FILL OUT HERE
    stack = ListStack()
    queue = ListQueue()

    for ch in A:
        stack.push(ch)
        queue.enqueue(ch)

    while not stack.isEmpty():
        if stack.pop() != queue.dequeue():
            return False

    return True

str = 'civic'
print(str, "is Palindrome?: ", isPalindrome(str))
```

## Stack: Webbrowser

```python
import webbrowser
import time
import sys
stack = ListStack()
urls = [ "naver.com", "daum.net", "nate.com"]
# Browser history stacks up in the order in which you visit websites
# push url of each visited websites to a stack
for url in urls:
  stack.push(url)
  # When you hit the back button, the previous URL is popped from the stack.

while not stack.isEmpty(): # if stack is not empty
  url = stack.pop() # pop the previous URL and open the browser
  webbrowser.open('http://'+url)
```
