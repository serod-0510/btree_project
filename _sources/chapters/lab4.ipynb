# Example 1: Time Complexity

What is the time complexity of access, search, insertion, deletion of array data structure? Explain why.


```python
import numpy as np

class MyArray:
    def __init__(self, input_list):
        self.array = np.array(input_list, dtype="i")

    def myInsert(self, x, i):
        new_array = np.empty(len(self.array) + 1, dtype="i")
        new_array[:len(self.array)] = self.array
        n = len(new_array)

        # Shift elements to the right
        for idx in range(n - 2, i - 1, -1):
            new_array[idx + 1] = new_array[idx]

        new_array[i] = x
        self.array = new_array  # Update self.array
        return self.array

    def myIndex(self, x):
        n = len(self.array)
        for idx in range(n):
            if self.array[idx] == x:
                return idx
        return -1  # If the key is not found

    def myDelete(self, i):
        n = len(self.array)

        # Shift elements to the left
        for idx in range(i + 1, n):
            self.array[idx - 1] = self.array[idx]

        self.array = self.array[:n - 1]  # Update self.array
        return self.array

    def myRemove(self, x):
        i = self.myIndex(x)
        if i == -1:
            print("Element not found.")
            return self.array

        print("Element found at position:", i)
        self.myDelete(i)
        return self.array
```

# Example 2: Python List vs Array

## Storage Structure

Python lists store references to objects, allowing mixed data types.
Although the references themselves are stored contiguously, the actual objects may be located in different memory locations.
The actual objects may be stored in different memory locations.

Arrays store elements of the same data type in contiguous memory locations, which improves memory efficiency and cache performance.

## Performance

Both lists and arrays support O(1) indexing.

| Operation | List | Array |
|-----------|------|--------|
| Access | O(1) | O(1) |
| Search | O(n) | O(n) |
| Insert | O(n) | O(n) |
| Delete | O(n) | O(n) |

Arrays are generally faster for numerical computations because they use contiguous memory and homogeneous data types.

## Type of Elements

Lists can store mixed data types, while arrays require elements of the same type.

## Dynamic vs Static

Python lists are dynamic arrays and automatically resize when needed. Arrays are generally less flexible and may require reallocation when resized.

## Access and Manipulation

Both support indexing and iteration, but lists provide more flexibility while arrays offer better performance for numerical operations.



# Example 3

Create a function that output unique elements in a given array. Unique means distinct elements. What is the time complexity of the algorithm?


```python
def unique(arr):
    unique_elements = []
    for element in arr:
        if element not in unique_elements:
            unique_elements.append(element)
    return unique_elements

a = [1, 2, 2, 3, 4, 5, 5, 6, 7, 7]
unique(a)
```

# Example 4

Create a function that returns how many number of elements are odd and even elements in the array.


```python
def EvenOdd(arr):
    even = 0
    odd = 0
    for element in arr:
        if element % 2 == 0:
            even += 1
        else:
            odd += 1
    print("Even:", even)
    print("Odd:", odd)

a = [1, 2, 2, 3, 4, 5, 5, 6, 7, 7]
EvenOdd(a)
```
